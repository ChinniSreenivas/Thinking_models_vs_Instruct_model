# 🤖 Models Used in the Project

This project compares four Large Language Models (LLMs) to understand the differences between **Thinking Models** and **Instruct Models**.

We use two models from each category.

---

## 1. DeepSeek-R1

**Type:** Thinking Model  
**Developer:** DeepSeek AI

DeepSeek-R1 is a reasoning-focused model designed to work through problems step by step before producing an answer. It is particularly suitable for tasks that require mathematical reasoning, logic, and problem-solving.

**Key characteristics:**
- Thinking/reasoning-focused model
- Designed for mathematical and logical problems
- Produces reasoning before the final answer
- Available through DeepSeek's platform/API

**Role in this project:**  
DeepSeek-R1 is used as a representative **Thinking Model**.

---

## 2. o3-mini

**Type:** Thinking Model  
**Developer:** OpenAI

o3-mini is a compact reasoning model from OpenAI. It is designed to solve problems that require reasoning while allowing different levels of reasoning effort.

**Key characteristics:**
- Reasoning-focused model
- Suitable for mathematics, logic, and problem-solving
- Supports different reasoning effort levels
- Developed by OpenAI

**Role in this project:**  
o3-mini is used as the second **Thinking Model** for comparison.

---

## 3. GPT-4o mini

**Type:** Instruct Model  
**Developer:** OpenAI

GPT-4o mini is a smaller, efficient model designed for fast and reliable responses to everyday tasks such as question answering and writing.

**Key characteristics:**
- Fast response generation
- Designed for everyday question answering
- Useful for writing and general-purpose tasks
- Uses fewer resources compared with larger models

**Role in this project:**  
GPT-4o mini is used as a representative **Instruct Model**.

---

## 4. Llama-3.1 8B Instruct

**Type:** Instruct Model  
**Developer:** Meta AI

Llama-3.1 8B Instruct is an instruction-following model from Meta AI. It is designed to respond directly to user instructions and can be used for general-purpose question answering and other language tasks.

**Key characteristics:**
- Instruction-following model
- Open-source model
- 8-billion-parameter model
- Suitable for general-purpose tasks
- Can be run through available AI platforms or locally

**Role in this project:**  
Llama-3.1 8B Instruct is used as the second **Instruct Model**.

```python
from huggingface_hub import InferenceClient

client = InferenceClient(
    model="meta-llama/Llama-3.1-8B-Instruct",
    token=HF_TOKEN
)

print("Llama-3.1 8B Instruct client created successfully!")
```

```python
question = "What is the capital of India?"

# Fix: Strip leading/trailing whitespace from the HF_TOKEN
cleaned_hf_token = HF_TOKEN.strip()

# Re-initialize the client with the corrected token
from huggingface_hub import InferenceClient
client = InferenceClient(
    model="meta-llama/Llama-3.1-8B-Instruct",
    token=cleaned_hf_token
)

response = client.chat_completion(
    messages=[
        {
            "role": "user",
            "content": question
        }
    ],
    max_tokens=100
)

answer = response.choices[0].message.content

print("Question:")
print(question)

print("\nLlama-3.1 8B Instruct Answer:")
print(answer)
```

```python
import time

question = "What is the capital of India?"

start_time = time.time()

response = client.chat_completion(
    messages=[
        {
            "role": "user",
            "content": question
        }
    ],
    max_tokens=100
)

end_time = time.time()

answer = response.choices[0].message.content
response_time = end_time - start_time

print("Question:", question)
print("\nAnswer:", answer)
print("\nResponse time:", round(response_time, 3), "seconds")
```

```python
questions = [
    "What is the capital of India?",

    "If a train travels 60 kilometers in 1 hour, how far will it travel in 3 hours?",

    "Maria has 5 apples. She gives 2 apples to her friend and buys 3 more. How many apples does she have now?",

    "What is the difference between supervised learning and unsupervised learning?",

    "Write a Python function that checks whether a number is even or odd.",

    "What is the time complexity of searching for an element in a Python list?",

    "If 5 workers can complete a task in 10 days, assuming everyone works at the same rate, how many days would 10 workers take?",

    "Explain what an API is in simple words.",

    "What are the main differences between RAM and ROM?",

    "A shop gives a 20% discount on an item priced at ₹1,000. What is the final price after the discount?"
]
```

```python
import time

results = []

for i, question in enumerate(questions, start=1):

    start_time = time.time()

    response = client.chat_completion(
        messages=[
            {
                "role": "user",
                "content": question
            }
        ],
        max_tokens=300
    )

    end_time = time.time()

    answer = response.choices[0].message.content
    response_time = end_time - start_time

    results.append({
        "Question_No": i,
        "Question": question,
        "Model": "Llama-3.1-8B-Instruct",
        "Answer": answer,
        "Response_Time_Seconds": round(response_time, 3)
    })

    print(f"\n{'='*60}")
    print(f"Question {i}: {question}")
    print(f"Response time: {response_time:.3f} seconds")
    print(f"Answer: {answer}")
```

```python
import pandas as pd

df = pd.DataFrame(results)

df
```
