```python
import os
import time
import pandas as pd
from openai import OpenAI
from huggingface_hub import login
from transformers import pipeline
```

```python
API_KEY = "add key"

client = OpenAI(api_key=API_KEY)




questions = [
    "What is the capital of India?",

    "If a train travels 60 kilometers in 1 hour, how far will it travel in 3 hours?",

    "Maria has 5 apples. She gives 2 apples to her friend and buys 3 more. How many apples does she have now?",

    "What is the difference between supervised learning and unsupervised learning?",

    "Write a Python function that checks whether a number is even or odd.",

    "What is the time complexity of searching for an element in a Python list?",

    "If 5 workers can complete a task in 10 days, assuming everyone works at the same rate, how many days would 10 workers take?",

    "Explain what an API is in simple words.",

    "What are the main differences between RAM and ROM?",

    "A shop gives a 20% discount on an item priced at ₹1,000. What is the final price after the discount?"
]
```

```python


#o3_mini

def ask_o3_mini(question):

    start_time = time.perf_counter()

    response = client.responses.create(
        model="o3-mini",
        input=question
    )

    end_time = time.perf_counter()

    response_time = end_time - start_time

    answer = response.output_text

    return answer, response_time




results = []

for i, question in enumerate(questions, start=1):

    print(f"Processing Question {i}/{len(questions)}...")

    try:
        answer, response_time = ask_o3_mini(question)

        results.append({
            "Question_No": i,
            "Question": question,
            "Model": "o3-mini",
            "Answer": answer,
            "Response_Time_Seconds": round(response_time, 3)
        })

        print(f"Completed in {response_time:.3f} seconds")

    except Exception as e:

        results.append({
            "Question_No": i,
            "Question": question,
            "Model": "o3-mini",
            "Answer": f"ERROR: {str(e)}",
            "Response_Time_Seconds": None
        })

        print(f"Error: {e}")
```

```python
df = pd.DataFrame(results)



output_file = "o3_mini_results.csv"

df.to_csv(output_file, index=False, encoding="utf-8")

print("\nDone!")
print(f"CSV saved as: {output_file}")



display(df)
```

```python
hf_token = ""

if hf_token is None:
    raise ValueError("HF_TOKEN not found. Add your Hugging Face token to Colab Secrets.")

login(token=hf_token)

model_name = "deepseek-ai/DeepSeek-R1-Distill-Qwen-7B"

pipe = pipeline(
    "text-generation",
    model=model_name,
    device_map="auto",
    dtype="auto"
)


def ask_deepseek(prompt):

    start_time = time.perf_counter()

    response = pipe(
        prompt,
        max_new_tokens=300,
        temperature=0.6,
        do_sample=True
    )

    end_time = time.perf_counter()

    response_time = end_time - start_time

    answer = response[0]["generated_text"]

    return answer, response_time


results2 = []

for i, question in enumerate(questions, start=1):

    print(f"Processing Question {i}/{len(questions)}...")

    try:

        answer, response_time = ask_deepseek(question)

        results2.append({
            "Question_No": i,
            "Question": question,
            "Model": "DeepSeek-R1-Distill-Qwen-7B",
            "Answer": answer,
            "Response_Time_Seconds": round(response_time, 3)
        })

        print(f"Response time: {response_time:.3f} seconds")
        print("-" * 60)

    except Exception as e:

        results2.append({
            "Question_No": i,
            "Question": question,
            "Model": "DeepSeek-R1-Distill-Qwen-7B",
            "Answer": f"ERROR: {str(e)}",
            "Response_Time_Seconds": None
        })

        print(f"Error: {e}")

```

```python
df = pd.DataFrame(results2)

output_file = "deepseek_r1_results.csv"

df.to_csv(
    output_file,
    index=False,
    encoding="utf-8-sig"
)


print("All 10 questions completed!")

display(df)
```
